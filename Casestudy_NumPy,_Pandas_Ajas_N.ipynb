```python
#Task 1    Import pandas, numpy

import pandas as pd
import numpy as np
```

```python
#Task 2    Load dataset using pd.read_csv()

filepath='https://raw.githubusercontent.com/laxmimerit/All-CSV-ML-Data-Files-Download/master/IMDB-Movie-Data.csv'
df=pd.read_csv(filepath)
df
```

```python
#Task 4    Check and remove duplicates

# to check it have
df.duplicated()
```

```python
# To remove duplicates
df.drop_duplicates()
```

```python
#Task 5      To rename the columns

df = df.rename(columns={"Title": "Movie_Title"})
```

```python
#Task 6      Get unique genres

df["Genre"].unique()
```

```python
#Task 7    Extract number of genres per movie

df["Genre_Count"] = df["Genre"].str.count(",") + 1
print(df[["Movie_Title", "Genre_Count"]])
```

```python
#Task 8   Handle missing values in 'Revenue (Millions)' and 'Metascore

print(df[["Revenue (Millions)", "Metascore"]].fillna(df[["Revenue (Millions)", "Metascore"]].mean()))
```

```python
#Task 9    Top 5 movies by Rating

print(df.nlargest(5, "Rating")[["Movie_Title", "Rating"]])
```

```python
#Task 10    Movies with highest/lowest revenue

print(df["Revenue (Millions)"].max())
print(df["Revenue (Millions)"].min())
```

```python
#Task 11    Avg. rating by Director

print(df.groupby("Director")["Rating"].mean())
```

```python
#Task 12    Movies per year

print(df["Year"].value_counts())
```

```python
#Task 13    Movies with rating > 8 & revenue > 100M

print(df[(df["Rating"] > 8) & (df["Revenue (Millions)"] > 100)]["Movie_Title"])
```

```python
#Task 14    Group by Director: avg Rating, Revenue

print(df.groupby("Director")[["Rating", "Revenue (Millions)"]].mean())
```

```python
#Task 15      Group by Year: count, avg Rating

print(df.groupby("Year")["Rating"].agg(["count", "mean"]))
```

```python
#Task 16     Create a new column Profitability:
#A) Revenue divided by Duration

df["Profitability"] = df["Revenue (Millions)"] / df["Runtime (Minutes)"]
```

```python
#B) Find top 10 most profitable movies

print(df.nlargest(10, "Profitability")[["Movie_Title", "Profitability"]])
```

```python
#Task 17     Create a column that categorizes movies by rating:
#A) Rating < 6: Poor

print(df[df["Rating"] < 6][["Movie_Title", "Rating"]])
```

```python
#B)  6 ≤ Rating < 7.5: Average

print(df[(df["Rating"] >= 6) & (df["Rating"] < 7.5)][["Movie_Title", "Rating"]])
```

```python
#C) Rating ≥ 7.5: Excellent

print(df[df["Rating"] >= 7.5][["Movie_Title", "Rating"]])
```

```python
#Task 18   Create a custom function to score movies based on:

#i) Apply row['Rating'] * 10 + row['Revenue (Millions)'] as High score Apply a function to dataset to assign the score

def score(row):
    return row["Rating"] * 10 + row["Revenue (Millions)"]

df["Score"] = df.apply(score, axis=1)

print(df[["Movie_Title", "Score"]])
```
